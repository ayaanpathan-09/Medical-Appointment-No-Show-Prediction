# 01 - Exploratory Data Analysis

Explores the historical appointment dataset before feature engineering.
Swap `DATA_PATH` for your real historical export (see `src/data/ingestion.py`
for the expected column names).

```python
import sys
sys.path.append('..')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.data.ingestion import load_appointments_csv
from src.data.preprocessing import clean_data

sns.set_theme(style='whitegrid')
```

```python
DATA_PATH = '../tests/test_data/sample_appointments.csv'  # replace with your real dataset

valid_df, audit = load_appointments_csv(DATA_PATH, training=True)
print(audit)
valid_df.head()
```

```python
df = clean_data(valid_df)
df.info()
```

## No-show class balance

```python
df['no_show'].value_counts(normalize=True).plot(kind='bar', title='No-Show Rate')
plt.ylabel('Proportion')
plt.show()
```

## Age distribution

```python
df['age'].hist(bins=20)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.show()
```

## No-show rate by gender and SMS status

```python
for col in ['gender', 'sms_received']:
    print(df.groupby(col)['no_show'].mean())
    print()
```
